```python
import numpy as np
import pandas as pd


# Number of repeats per event
N = 1

# Photon counts you want to generate
#n_values = np.arange(1, 160, 2, dtype=int)   # 2, 12, 22, ..., 202
n_values = np.linspace(0.25,25,11)
print(n_values)
# Fixed values
x = y  = t = 0.0
dx = dy = dz = dt = 0.0
z=1390/2
p = 9.7
dp = 0.1

rows = []

# Repeat each event N times
for n in n_values:
    for _ in range(N):
        rows.append([x, y, z, t, dx, dy, dz, dt, p, dp, n*1e6])

# Convert to DataFrame
df = pd.DataFrame(rows, columns=[
    'x','y','z','t','dx','dy','dz','dt','p','dp','n'
])

# Save to file
filename="../txtfiles/Performance.txt"
df.to_csv(filename, sep="\t", index=False)

print("Generated", len(df), "rows")

# Save to file
#df.to_csv("txtfiles/HD_Performance.txt", sep="\t", index=False)

print(f"Generated {filename} with", N, "entries")
```
